```python
import gudhi
import numpy as np
import galois
```

torus z zadania 3 z lab2

```python
torus = gudhi.SimplexTree()
for sigma in [[i, (i + 1) % 7, (i + 3) % 7] for i in range(7)] + \
             [[i, (i + 2) % 7, (i + 3) % 7] for i in range(7)]:
    torus.insert(sigma)

print(f'torus: f = {torus.num_simplices_by_dimension()}, chi = {torus.euler_characteristic()}')
```

liczby Bettiego z gotowej funkcji

```python
torus.compute_persistence(homology_coeff_field=2, persistence_dim_max=True)

print(f'liczby Bettiego: {torus.betti_numbers()}')
```

macierze brzegu

```python
def simplices(st, k):
    return sorted(tuple(s) for s, _ in st.get_skeleton(k) if len(s) == k + 1)

def boundary_matrix(st, k):
    index = {tau: i for i, tau in enumerate(simplices(st, k - 1))}
    cols = simplices(st, k)
    D = np.zeros((len(index), len(cols)), dtype=int)
    for j, sigma in enumerate(cols):
        for tau, _ in st.get_boundaries(list(sigma)):
            D[index[tuple(tau)], j] = 1
        return D
```

f

```python
D = torus.dimension()
f = [len(simplices(torus, k)) for k in range(D + 1)]
```

r

```python
GF2 = galois.GF(2)

def r(st, k):
    if k <= 0 or k > st.dimension():
        return 0
    return np.linalg.matrix_rank(GF2(boundary_matrix(st, k)))
```

liczby Bettiego

```python
def beta(i):
    return f[i] - r(torus, i) - r(torus, i+1)

print(beta(2))
```
