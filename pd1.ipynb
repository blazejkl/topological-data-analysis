```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from scipy.spatial.distance import pdist
```

##### **zadanie 1**

```python
def sfera(n, k):
    A = np.random.randn(n, k+1)
    return A / np.linalg.norm(A, axis=1, keepdims=True)
```

```python
plt.plot()
plt.hist(sfera(10000, 2)[:, 2], bins=10)
plt.show()
```

##### **zadanie 3**

```python
def zanurz(X, d):
    m = X.shape[1]
    A, _ = np.linalg.qr(np.random.randn(d, m))
    return np.matmul(X, A.T) 

X = sfera(200, 1)
D = pdist(X)
for d in [2, 10, 100, 1000]:
    Y = zanurz(X, d)
    err = np.abs(pdist(Y) - D).max()
    print(d, Y.shape, f"max |Δ| = {err:.2e}", np.allclose(pdist(Y), D))
```

##### **zadanie 4**

```python
n = 400
k = 1 
sigma = 0.1
D = [2, 10, 100, 1000]
pdist_1 = [] # wariant 1
pdist_2 = [] # wariant 2
pdist_3 = [] # wariant 3

for d in D:
    m_j = zanurz(sfera(n, k), d)
    
    e_j1 = np.random.normal(0, sigma, size=(n, d))
    e_j2 = np.random.normal(0, sigma/np.sqrt(d), size=(n, d))
    e_j3 = np.random.normal(0, sigma, size=(n, d))

    pdist_1.append(pdist(m_j + e_j1))
    pdist_2.append(pdist(m_j + e_j2))
    pdist_3.append(pdist(e_j3))

cv_1 = np.std(pdist_1, axis=1) / np.mean(pdist_1, axis=1)
cv_2 = np.std(pdist_2, axis=1) / np.mean(pdist_2, axis=1)
cv_3 = np.std(pdist_3, axis=1) / np.mean(pdist_3, axis=1)

print(f'cv1: {cv_1}')
print(f'cv2: {cv_2}')
print(f'cv3: {cv_3}')

```
