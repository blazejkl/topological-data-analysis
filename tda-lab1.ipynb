```python
def sfera(n, k):
    A = np.random.randn(n, k+1)
    return A / np.linalg.norm(A, axis=1, keepdims=True)
```

```python
plt.plot()
plt.hist(sfera(10000, 2)[:, 2], bins=10)
plt.show()
```
