````python
import gudhi
import numpy as np
import galois
````

Macierz brzegu z parametrem p

````python
def simplices(st, k):
    return sorted(tuple(s) for s, _ in st.get_skeleton(k) if len(s) == k + 1)

def boundary_matrix(st, k, p=2):
    cols = simplices(st, k)
    rows = simplices(st, k - 1)
    index = {tau: i for i, tau in enumerate(rows)}
    D = np.zeros((len(index), len(cols)), dtype=int)

    for j, sigma in enumerate(cols):
        for i in range(len(sigma)):
            tau = sigma[:i] + sigma[i + 1:]
            D[index[tau], j] = (-1) ** i % p
    return D
````

Funkcja f

````python
def f(sk, i):
    return len(simplices(sk, i))
````

Funkcja r

````python
def r(st, k, p=2):
    if k <= 0 or k > st.dimension():
        return 0
    GF2 = galois.GF(p)
    return np.linalg.matrix_rank(GF2(boundary_matrix(st, k, p)))
````

##### **Zadanie 1**

Funkcja Bettiego

````python
def betti(sk, p=2):
    return [f(sk, i) - r(sk, i, p) - r(sk, i+1, p) for i in range(3)]
````

Torus z lab2

````python
torus = gudhi.SimplexTree()
for sigma in [[i, (i + 1) % 7, (i + 3) % 7] for i in range(7)] + \
             [[i, (i + 2) % 7, (i + 3) % 7] for i in range(7)]:
    torus.insert(sigma)

print(f'torus: f = {torus.num_simplices_by_dimension()}, chi = {torus.euler_characteristic()}')
print(f'Betti dla torusa nad Z2: {betti(torus, p=2)}')
print(f'Betti dla torusa nad Z3: {betti(torus, p=3)}')
````

podsumowanie wszystkich kształtów i sprawdzenie Eulera-Poincarego w kodzie

````python
ksztalty = {'sfera': sfera, 'torus': torus, 'butelka Kleina': klein,
            'RP2': rp2, 'wstega Mobiusa': mobius, 'walec': walec}

for nazwa, K in ksztalty.items():
    chi = K.euler_characteristic()
    b2, b3 = betti(K, p=2), betti(K, p=3)
    ep2 = chi == b2[0] - b2[1] + b2[2]
    ep3 = chi == b3[0] - b3[1] + b3[2]
    print(f'{nazwa:15} chi = {chi:2}   Z2: {b2}   Z3: {b3}   E-P: {ep2}, {ep3}')
````

**Wnioski do zadania 1**

- twierdzenie Eulera-Poincarego zgadza się dla wszystkich 6 kształtów, nad Z2 i nad Z3. $\chi$ liczymy z samego wektora $f$, więc nie zależy od ciała, a suma naprzemienna liczb Bettiego wychodzi ta sama nawet wtedy, gdy same liczby Bettiego się różnią
- dla sfery, torusa, walca i wstęgi Möbiusa liczby Bettiego nad Z2 i Z3 są takie same
- różnią się tylko dla butelki Kleina i $\mathbb{R}P^2$, czyli zamkniętych powierzchni nieorientowalnych:
    - nad Z2 suma wszystkich trójkątów jest cyklem, bo każda krawędź leży w dwóch trójkątach i w brzegu pojawia się dwa razy, a $1 + 1 = 0$. stąd $\beta_2 = 1$
    - nad Z3 znaki mają znaczenie. współczynnik przy jednym trójkącie wymusza współczynniki u sąsiadów (z dokładnością do znaku), więc niezerowy cykl to tak naprawdę zgodne zorientowanie wszystkich trójkątów naraz. na powierzchni nieorientowalnej się tego nie da, więc $\beta_2 = 0$
    - $\chi$ się nie zmienia, więc skoro $\beta_2$ spada o 1, to $\beta_1$ też musi spaść o 1 (Klein: 2 → 1, RP2: 1 → 0)
    - z tego co wiem, nad $\mathbb{Z}$ jest $H_1(\text{Klein}) = \mathbb{Z} \oplus \mathbb{Z}_2$ i $H_1(\mathbb{R}P^2) = \mathbb{Z}_2$. ten kawałek $\mathbb{Z}_2$ widać nad Z2, a nad Z3 znika
- nad samym Z2 torus i butelka Kleina mają te same liczby Bettiego $(1, 2, 1)$, więc Z2 ich nie rozróżnia, dopiero Z3
- walec i wstęga Möbiusa mają $(1, 1, 0)$ nad obydwoma ciałami, więc liczby Bettiego ich nie rozróżniają (oba ściągają się do okręgu). $\beta_2 = 0$, bo mają brzeg: krawędź brzegowa leży w jednym trójkącie, więc ten trójkąt musi mieć w cyklu współczynnik 0, potem jego sąsiedzi itd., czyli jedyny 2-cykl to zero
- wynik nie zależy od triangulacji (torus ma 7 wierzchołków, butelka Kleina 9, sfera 14), ale triangulacja musi być poprawna: przy RP2 widać, że warto sprawdzać, czy $f_2$ = 2 · liczba kwadracików

##### **Zadanie 2**

Funkcja sprawdzająca, czy kompleks jest zamkniętą powierzchnią

````python
def zamknieta_powierzchnia(K):
    # 1. kazda krawedz lezy w dokladnie dwoch trojkatach
    for e in simplices(K, 1):
        n = len(K.get_cofaces(list(e), 1))
        if n != 2:
            print(f'  krawedz {e} nie lezy w dokladnie 2 trojkatach (lezy w {n})')
            return False

    # 2. link kazdego wierzcholka jest cyklem prostym
    for (v,) in simplices(K, 0):
        link = gudhi.SimplexTree()
        for t, _ in K.get_cofaces([v], 2):      # trojkaty zawierajace v
            link.insert([u for u in t if u != v])
        stopnie = [len(link.get_cofaces([u], 1)) for (u,) in simplices(link, 0)]
        # cykl prosty = kazdy wierzcholek stopnia 2 i graf spojny
        if any(d != 2 for d in stopnie) or betti(link)[0] != 1:
            print(f'  link wierzcholka {v} nie jest cyklem prostym')
            return False

    # 3. K jest spojny
    if betti(K)[0] != 1:
        print('  kompleks nie jest spojny')
        return False

    return True
````

sprawdzenie triangulacji z zadania 1

````python
for nazwa, K in ksztalty.items():
    print(f'{nazwa}: {zamknieta_powierzchnia(K)}')
````
