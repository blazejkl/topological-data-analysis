##### **zadanie 1a**

**okrąg**: 

$V$ = {a, b, c}

$K$ = {{a}, {b}, {c}, {a, b}, {b, c}, {a, c}}

$f$ = [3 3]

$\chi(K) = 1*3 - 1*3 = 0$


**ósemka**:

a - punkt styku; b, c - pierwsze kółko; d, e - drugie kółko

$V$ = {a, b, c, d, e}

$K$ = {{a}, {b}, {c}, {d}, {e}, {a, b}, {b, c}, {a, c}, {a, d}, {d, e}, {a, e}}

$f$ = [5 6]

$\chi(K) = 5 - 6 = -1$


**trzy okręgi**:

a, b, c - punkty styku; d, e, f - dodatkowe wierzchołki

$V$ = {a, b, c, d, e, f}

$K$ = {{a}, ..., {f}, 
        {a, b}, {a, c}, {b, c}, 
        {a, d}, {b, d},
        {a, e}, {c, e},
        {b, f}, {c, f}}

$f$ = [6 9]

$\chi(K) = 6 - 9 = -3$


implementacja tych kompleksów:

````python
okrag = gudhi.SimplexTree()
for sigma in [[0, 1], [1, 2], [0, 2]]:
    okrag.insert(sigma)

osemka = gudhi.SimplexTree()
for sigma in [[0, 1], [1, 2], [0, 2],
                [0, 3], [3, 4], [0, 4]]:
    osemka.insert(sigma)

trzy_okregi = gudhi.SimplexTree()
for sigma in [[0, 1], [1, 2], [0, 2],
               [0, 3], [1, 3],
               [0, 4], [2, 4],
               [1, 5], [2, 5]]:
    trzy_okregi.insert(sigma)

````

* f-wektor: ```st.num_simplices_by_dimension()``` 
* $\chi(K)$: ```st.euler_characteristic()``` 

````python
print(f'okrąg: f = {okrag.num_simplices_by_dimension()}, chi = {okrag.euler_characteristic()}')
print(f'ósemka: f = {osemka.num_simplices_by_dimension()}, chi = {osemka.euler_characteristic()}')
print(f'trzy okręgi: f = {trzy_okregi.num_simplices_by_dimension()}, chi = {trzy_okregi.euler_characteristic()}')
````

##### **zadanie 2b**

Przykładowy sposób otrzymania triangulacji sfery o dowolnie dużej liczbie wierzchołków z wykorzystaniem obiektów z zadania 1 - konstruujemy siatkę południków i równoleżników:
* dwa bieguny: N i S
* k równoleżników - każdy to cykl z m wierzchołków, czyli okrąg z zadania 1 
* między równoleżnikami pasek, czyli pierścień z 1c
* N łączymy z każdą krawędzią pierwszego równoleżnika, S - ostatniego (po m trójkątów)
* liczba wierzchołków to $k * m + 2$

````python
# numeracja: biegun N = 0, równoleżnik j (j = 0..k-1) ma wierzchołki 1 + j*m + i (i = 0..m-1), biegun S = k*m + 1
rozmiary = [(1, 3), (2, 4), (5, 10), (20, 50), (100, 1000)]

for k, m in rozmiary:
    N = 0
    S = k * m + 1

    czapka_N = [[N, 1 + i, 1 + (i + 1) % m] for i in range(m)]
    czapka_S = [[S, 1 + (k - 1)*m + i, 1 + (k - 1)*m + (i + 1) % m] for i in range(m)]

    # paski między równoleżnikiem j i j+1: ten sam wzór co pierścień z 1c
    paski = [[1 + j*m + i, 1 + j*m + (i + 1) % m, 1 + (j + 1)*m + (i + 1) % m] for j in range(k - 1) for i in range(m)] + \
            [[1 + j*m + i, 1 + (j + 1)*m + (i + 1) % m, 1 + (j + 1)*m + i] for j in range(k - 1) for i in range(m)]

    sfera_2b = gudhi.SimplexTree()
    for sigma in czapka_N + czapka_S + paski:
        sfera_2b.insert(sigma)

    print(f'k = {k}, m = {m}: f = {sfera_2b.num_simplices_by_dimension()}, chi = {sfera_2b.euler_characteristic()}')
````

##### **zadanie 2c**

weźmy gęstą triangulację z 2b dla k = 20, m = 50 i usuńmy po kolei trzy trójkąty z różnych pasków

````python
k, m = 20, 50
N = 0
S = k * m + 1

czapka_N = [[N, 1 + i, 1 + (i + 1) % m] for i in range(m)]
czapka_S = [[S, 1 + (k - 1)*m + i, 1 + (k - 1)*m + (i + 1) % m] for i in range(m)]
paski = [[1 + j*m + i, 1 + j*m + (i + 1) % m, 1 + (j + 1)*m + (i + 1) % m] for j in range(k - 1) for i in range(m)] + \
        [[1 + j*m + i, 1 + (j + 1)*m + (i + 1) % m, 1 + (j + 1)*m + i] for j in range(k - 1) for i in range(m)]

sfera_2c = gudhi.SimplexTree()
for sigma in czapka_N + czapka_S + paski:
    sfera_2c.insert(sigma)

print(f'sfera: f = {sfera_2c.num_simplices_by_dimension()}, chi = {sfera_2c.euler_characteristic()}')

# trzy trójkąty parami bez wspólnych wierzchołków - [101, 102, 152], [401, 402, 452], [701, 702, 752]

sfera_2c.remove_maximal_simplex([101, 102, 152])
print(f'bez 1 trójkąta: f = {sfera_2c.num_simplices_by_dimension()}, chi = {sfera_2c.euler_characteristic()}')

sfera_2c.remove_maximal_simplex([401, 402, 452])
print(f'bez 2 trójkątów: f = {sfera_2c.num_simplices_by_dimension()}, chi = {sfera_2c.euler_characteristic()}')

sfera_2c.remove_maximal_simplex([701, 702, 752])
print(f'bez 3 trójkątów: f = {sfera_2c.num_simplices_by_dimension()}, chi = {sfera_2c.euler_characteristic()}')
````

Wnioski: wartości $\chi$ w 2a i 2b potwierdzają hipotezę o równym $\chi$ dla kształtów homotopijnie równoważnych. 

Usunięcie każdego trójkąta obniżało $\chi$ o jeden, np. dla dwóch usuniętych mieliśmy wartość 0, zatem tyle samo co dla okręgu i pogrubionego okręgu w zadaniu 1. Zgadza się to z hipotezą, bo sfera bez dwóch trójkątów to topologicznie walec, który ściąga się do okręgu. 

Analogicznie bez jednego trójkąta dostajemy dysk, a bez trzech dysk z dwiema dziurami, czyli pogrubioną ósemkę.

##### **zadanie 3**

wykorzystajmy 'torus Möbiusa-Császára', czyli minimalną możliwą triangulację:

wierzchołki $0, \dots, 6$, trójkąty $\{i, i+1, i+3\}$ oraz $\{i, i+2, i+3\}$ (mod 7) dla $i = 0, \dots, 6$

$V$ = {a, b, c, d, e, f, g}

$K$ = {wszystkie wierzchołki, wszystkie 21 par wierzchołków,
    {a,b,d}, {a,b,f}, {a,c,d}, {a,c,g}, {a,e,f}, {a,e,g}, {b,c,e}, 
    {b,c,g}, {b,d,e}, {b,f,g}, {c,d,f}, {c,e,f}, {d,e,g}, {d,f,g}}

$f$ = [7 21 14]

$\chi(K) = 7 - 21 + 14 = 0$

````python
torus = gudhi.SimplexTree()
for sigma in [[i, (i + 1) % 7, (i + 3) % 7] for i in range(7)] + \
             [[i, (i + 2) % 7, (i + 3) % 7] for i in range(7)]:
    torus.insert(sigma)

print(f'torus: f = {torus.num_simplices_by_dimension()}, chi = {torus.euler_characteristic()}')
````

**triangulacja o dowolnie dużej liczbie wierzchołków**

analogicznie do globusa z 2b - torus to okrąg × okrąg, więc bierzemy siatkę okręgów:
* k okręgów, każdy to cykl z m wierzchołków (okrąg z zadania 1)
* między sąsiednimi okręgami pasek, czyli pierścień z 1c
* zamiast biegunów sklejamy ostatni okrąg z pierwszym (indeks okręgu mod k)
* potrzeba $k, m \ge 3$, inaczej powstają powtórzone sympleksy
* liczba wierzchołków to $k \cdot m$, a $f = [km \;\; 3km \;\; 2km]$

````python
# numeracja: okrąg j (j = 0..k-1) ma wierzchołki j*m + i (i = 0..m-1), okrąg k utożsamiony z okręgiem 0
rozmiary = [(3, 3), (4, 5), (10, 10), (20, 50), (100, 1000)]   # (k, m)

for k, m in rozmiary:
    paski = [[j*m + i, j*m + (i + 1) % m, ((j + 1) % k)*m + (i + 1) % m] for j in range(k) for i in range(m)] + \
            [[j*m + i, ((j + 1) % k)*m + (i + 1) % m, ((j + 1) % k)*m + i] for j in range(k) for i in range(m)]

    torus_3b = gudhi.SimplexTree()
    for sigma in paski:
        torus_3b.insert(sigma)

    print(f'k = {k}, m = {m}: f = {torus_3b.num_simplices_by_dimension()}, chi = {torus_3b.euler_characteristic()}')
````

weźmy gęstą triangulację z 3b dla k = 20, m = 50 i usuńmy po kolei trzy trójkąty z różnych pasków

````python
k, m = 20, 50

paski = [[j*m + i, j*m + (i + 1) % m, ((j + 1) % k)*m + (i + 1) % m] for j in range(k) for i in range(m)] + \
        [[j*m + i, ((j + 1) % k)*m + (i + 1) % m, ((j + 1) % k)*m + i] for j in range(k) for i in range(m)]

torus_3c = gudhi.SimplexTree()
for sigma in paski:
    torus_3c.insert(sigma)

print(f'torus: f = {torus_3c.num_simplices_by_dimension()}, chi = {torus_3c.euler_characteristic()}')

# trzy trójkąty parami bez wspólnych wierzchołków - [100, 101, 151], [400, 401, 451], [700, 701, 751]

torus_3c.remove_maximal_simplex([100, 101, 151])
print(f'bez 1 trójkąta: f = {torus_3c.num_simplices_by_dimension()}, chi = {torus_3c.euler_characteristic()}')

torus_3c.remove_maximal_simplex([400, 401, 451])
print(f'bez 2 trójkątów: f = {torus_3c.num_simplices_by_dimension()}, chi = {torus_3c.euler_characteristic()}')

torus_3c.remove_maximal_simplex([700, 701, 751])
print(f'bez 3 trójkątów: f = {torus_3c.num_simplices_by_dimension()}, chi = {torus_3c.euler_characteristic()}')
````

**Wnioski**

Wyniki $\chi = 0$ dla torusa niezależnie od triangulacji potwierdzają hipotezę z zadania 1.

Analogicznie do zadania 2 usunięcie każdego trójkąta obniża $\chi$ o jeden. 

Czy kształty o tym samym $\chi$ muszą być "takie same"? Nie, kontrprzykłady:
* Torus i okrąg mają $\chi = 0$, ale nie są nawet homotopijnie równoważne. 
* Torus bez trójkąta i pogrubiona ósemka mają $\chi = -1$ i ten sam typ homotopijny, ale nie są homeomorficzne.
