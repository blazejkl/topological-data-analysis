```python
import gudhi
import numpy as np
import galois
import itertools
import matplotlib.pyplot as plt
from math import comb
from time import perf_counter
```

## **A**

```python
L = 2        
XA = np.loadtxt('lab6_A.csv', delimiter=',')
st_alpha = gudhi.AlphaComplex(points=XA).create_simplex_tree()
#rc = gudhi.RipsComplex(points=X, max_edge_length=L)
#st = rc.create_simplex_tree(max_dimension=3)
```

```python
paski2 = st_alpha.persistence(homology_coeff_field=2)
paski2_r = [(k, (np.sqrt(b), np.sqrt(d))) for k, (b, d) in paski2]
gudhi.plot_persistence_barcode(paski2_r, legend=True, max_intervals=50)
plt.xlabel('r')
plt.show()
```

```python
paski3 = st_alpha.persistence(homology_coeff_field=3)
paski3_r = [(k, (np.sqrt(b), np.sqrt(d))) for k, (b, d) in paski3]
gudhi.plot_persistence_barcode(paski3_r, legend=True, max_intervals=50)
plt.xlabel('r')
plt.show()
```

nad Z2 i Z3 mamy liczby bettiego (1, 2, 1), więc mamy torus

```python
L = 2        
rc = gudhi.RipsComplex(points=XA, max_edge_length=L)
st = rc.create_simplex_tree(max_dimension=3)
```

```python
L = 0.6
rc = gudhi.RipsComplex(points=XA[:600], max_edge_length=L)
st = rc.create_simplex_tree(max_dimension=3)
```

```python
print(f'sympleksy: {st.num_simplices()}')
print(f'f = {st.num_simplices_by_dimension()}')
```

```python
paski_rips = st.persistence(homology_coeff_field=2)
paski_rips_r = [(k, (b / 2, d / 2)) for k, (b, d) in paski_rips]
gudhi.plot_persistence_barcode(paski_rips_r, legend=True, max_intervals=50)
plt.title('A, 600 pkt, Rips, Z2')
plt.xlabel('r')
plt.show()
```
